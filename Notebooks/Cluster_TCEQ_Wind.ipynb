# Cluster wind data 

```python
import pandas as pd

# Load the processed wind data (which includes U, V winds)
processed_wind_data = pd.read_csv('processed_wind_data_1998_2024.csv')

# Load the clustered wind data file
clustered_wind_data = pd.read_csv('clustered_wind_data_1998_2024_k4.csv')

# Ensure 'Timestamp' in processed_wind_data is in datetime format and extract only the date
processed_wind_data['Timestamp'] = pd.to_datetime(processed_wind_data['Timestamp'])
processed_wind_data['date'] = processed_wind_data['Timestamp'].dt.date  # Extract only the date part

# Ensure 'Date' in clustered_wind_data is in datetime format and extract only the date
clustered_wind_data['Date'] = pd.to_datetime(clustered_wind_data['Date'])
clustered_wind_data['date'] = clustered_wind_data['Date'].dt.date  # Extract only the date part

# Merge the processed wind data with the cluster information
merged_df = pd.merge(processed_wind_data, clustered_wind_data[['date', 'Cluster']], on='date', how='left')

# Drop the temporary 'date' column after merging
merged_df.drop('date', axis=1, inplace=True)

# Split and save data based on Cluster values (0, 1, 2, 3)
for cluster in [0, 1, 2, 3]:
    cluster_df = merged_df[merged_df['Cluster'] == cluster]
    cluster_df.to_csv(f'processed_wind_data_cluster_{cluster}.csv', index=False)
    print(f"Saved: processed_wind_data_cluster_{cluster}.csv")

print("Processed wind data (U, V winds) has been split into separate CSV files for each cluster (0, 1, 2, 3).")
```

```python
import pandas as pd

# Define clusters
clusters = [0, 1, 2, 3]

for cluster in clusters:
    # Load the clustered wind data with U, V components
    cluster_df = pd.read_csv(f'processed_wind_data_cluster_{cluster}.csv')

    # Ensure 'Timestamp' is in datetime format
    cluster_df['Timestamp'] = pd.to_datetime(cluster_df['Timestamp'])

    # Extract the hour of the day
    cluster_df['hour'] = cluster_df['Timestamp'].dt.hour

    # Compute the composite (mean over the entire period) for wind speed, wind direction, U, and V
    composite_df = cluster_df.groupby('hour')[['Wind_Speed', 'Wind_Direction', 'U', 'V']].mean().reset_index()

    # Save the final composite file
    composite_df.to_csv(f'composite_processed_wind_cluster_{cluster}.csv', index=False)

    print(f"Saved: composite_processed_wind_cluster_{cluster}.csv")

print("Composite wind data (including U and V) has been generated for each cluster (0, 1, 2, 3).")
```
